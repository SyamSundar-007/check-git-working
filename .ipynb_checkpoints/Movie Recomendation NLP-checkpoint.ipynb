```python
df_new.isnull().sum()

df_new['overview'] = df_new['overview'].fillna('')
```

```python
#  extract genres from the text dict 
df_new['genres'] = df_new['genres'].apply( lambda x: " ".join( [  i['name'] for i in ast.literal_eval(x)   ] ))
```

```python
df_new
```

```python
df_new['tagline']= df_new['tagline'].fillna("")
```

```python
df_new.isnull().sum()
```

```python
df_new = df_new.dropna(subset=['title'])
```

```python
df_new.isnull().sum()
```

```python
df_new['tags'] = df_new['overview']+ " "+ df_new['genres']+" " + df_new['tagline']
```
